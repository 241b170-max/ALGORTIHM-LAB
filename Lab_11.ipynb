# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp_with_path(cost):
    n = len(cost)
    dp = [[float('inf')] * n for _ in range(1 << n)]
    parent = [[-1] * n for _ in range(1 << n)]

    dp[1][0] = 0  # start at city 0

    for mask in range(1 << n):
        for i in range(n):
            if dp[mask][i] == float('inf'):
                continue
            for j in range(n):
                if mask & (1 << j) == 0:  # city j not visited
                    new_mask = mask | (1 << j)
                    new_cost = dp[mask][i] + cost[i][j]
                    if new_cost < dp[new_mask][j]:
                        dp[new_mask][j] = new_cost
                        parent[new_mask][j] = i

    # Find best path returning to source
    full_mask = (1 << n) - 1
    min_cost = float('inf')
    last_city = -1
    for i in range(n):
        cost_to_return = dp[full_mask][i] + cost[i][0]
        if cost_to_return < min_cost:
            min_cost = cost_to_return
            last_city = i

    # Reconstruct path
    path = []
    mask = full_mask
    while last_city != -1:
        path.append(last_city)
        prev_city = parent[mask][last_city]
        mask = mask ^ (1 << last_city)
        last_city = prev_city

    path.append(0)  # return to source
    path.reverse()

    return min_cost, path
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```
